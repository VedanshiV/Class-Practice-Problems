```python
# 1. Create a 3x3 NumPy array of all True
# np.ones with dtype=bool fills every cell with True.
# (np.full((3, 3), True) works too.)
import numpy as np
print("1. 3x3 array of all True")
a1 = np.ones((3, 3), dtype=bool)
print(a1, "\n")
```

```python
# 2. Create a 1D array of 10 evenly spaced values between 5 and 50
# ---------------------------------------------------------------
# np.linspace(start, stop, num) includes both endpoints.
# Step = (50 - 5) / (10 - 1) = 5
print("2. 10 evenly spaced values between 5 and 50")
a2 = np.linspace(5, 50, 10)
print(a2, "\n")
```

```python
# 3. Convert a Python list into a NumPy array
# ---------------------------------------------------------------
# np.array() copies the list data into an ndarray.
print("3. Python list -> NumPy array")
py_list = [10, 20, 30, 40, 50]
a3 = np.array(py_list)
print("List :", py_list, type(py_list))
print("Array:", a3, type(a3), "\n")
```

```python
# 4. Reverse a 1D NumPy array
# ---------------------------------------------------------------
# Slicing with step -1 walks the array backwards.
# (np.flip(a4) gives the same result.)
print("4. Reverse a 1D array")
a4 = np.array([1, 2, 3, 4, 5])
print("Original:", a4)
print("Reversed:", a4[::-1], "\n")
```

```python
# 5. Create a 3x3 identity matrix
# ---------------------------------------------------------------
# np.eye(n) (or np.identity(n)) puts 1s on the main diagonal, 0s elsewhere.
print("5. 3x3 identity matrix")
a5 = np.eye(3)
print(a5, "\n")
```

```python
# 6. Create a 4x4 array; extract its first row and last column
# ---------------------------------------------------------------
# np.arange(1, 17) gives 1..16, reshape(4, 4) arranges it as 4 rows x 4 cols.
# Indexing is arr[row, col]; ':' means "all".
print("6. First row and last column of a 4x4 array")
a6 = np.arange(1, 17).reshape(4, 4)
print("Array:\n", a6)
print("First row  :", a6[0, :])     # row 0, all columns
print("Last column:", a6[:, -1], "\n")  # all rows, last column
```

```python
# 7. Create a 4x4 array; extract first two rows and first two columns
# ---------------------------------------------------------------
# Slice a[0:2, 0:2] -> rows 0-1 and columns 0-1 (stop index is excluded).
print("7. First two rows and first two columns")
a7 = np.arange(1, 17).reshape(4, 4)
print("Array:\n", a7)
print("First two rows   :\n", a7[:2, :])
print("First two columns:\n", a7[:, :2])
print("Top-left 2x2 block:\n", a7[:2, :2], "\n")
```

```python
# 8. Arithmetic operations on two NumPy arrays element-wise
# ---------------------------------------------------------------
# Operators act on matching positions; shapes must match (or broadcast).
print("8. Element-wise arithmetic")
x = np.array([10, 20, 30, 40])
y = np.array([1, 2, 3, 4])
print("x =", x, " y =", y)
print("Add          :", x + y)    # np.add(x, y)
print("Subtract     :", x - y)    # np.subtract(x, y)
print("Multiply     :", x * y)    # np.multiply(x, y)
print("Divide       :", x / y)    # np.divide(x, y)
print("Floor divide :", x // y)
print("Modulus      :", x % y)
print("Power        :", x ** y, "\n")
```

```python
# 9. Dot product of two NumPy arrays
# ---------------------------------------------------------------
# For 1D arrays: sum(a[i] * b[i]).
# 1*4 + 2*5 + 3*6 = 32
print("9. Dot product")
p = np.array([1, 2, 3])
q = np.array([4, 5, 6])
print("np.dot(p, q) :", np.dot(p, q))
print("p @ q        :", p @ q)

# For 2D arrays the same functions perform matrix multiplication.
m1 = np.array([[1, 2], [3, 4]])
m2 = np.array([[5, 6], [7, 8]])
print("Matrix product:\n", np.dot(m1, m2), "\n")
```

```python
# 10. Mean, median and standard deviation of a NumPy array
# ---------------------------------------------------------------
# mean   = sum / n
# median = middle value of the sorted data
# std    = sqrt(mean((x - mean)^2))  (population std, ddof=0 by default)
print("10. Mean, median, standard deviation")
data = np.array([4, 8, 15, 16, 23, 42])
print("Data  :", data)
print("Mean  :", np.mean(data))
print("Median:", np.median(data))
print("Std   :", np.std(data))
print("Sample std (ddof=1):", np.std(data, ddof=1))
```
