```python
# PANDAS
## A series is a one dimensional labelled as array capable of holding any data type
# example 1
import pandas as pd
a = [1,7,2]
myvar  = pd.Series(a)
print(myvar)
```

```python
# example 2 - name your own labels
import pandas as pd
a = [1,7,2]
myvar = pd.Series(a, index = ["x","y","z"])
print(myvar)
```

```python
# example 3
import pandas as pd
s1 = pd.Series([10,20,30] , index = ['a','b','c'] , name = "Example Series")
print(s1)
```

```python
# example 4
import pandas as pd
s2 = pd.Series({'x':1, 'y':2, 'z':3})
print(s2)
```

```python
# example 5
import pandas as pd
calories = {"day1": 420, "day2": 380, "day3": 390}
myvar = pd.Series(calories)
print(myvar)
```
