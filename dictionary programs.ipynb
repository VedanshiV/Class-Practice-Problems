```python
# question 4
my_dict = {
    101: "vedanshi",
    102: "aryaa",
    103: "aditya",
    104: "john"
}
print("keys: ")
for key in my_dict:
    print(key)
print("\nvalues:")
for value in my_dict.values():
    print(value)
print("\nkeys and values: ")
for key, value in my_dict.items():
    print(key,value)
```

```python
# question 6
my_dict = {"a":10, "b":20, "c":30, "d":40}
total = 0 
for value in my_dict.values():
 total = total + value
print("sum of values: ", total)
```
