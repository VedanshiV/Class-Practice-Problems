```python
# question 10
numbers = []
for i in range(10):
    num = int(input("Enter value: "))
    numbers.append(num)
print("\nOriginal List:", numbers)

# a) ascending order using sorted()
ascending = sorted(numbers)
print("\nAscending Order:", ascending)

# b) descending order using sort()
numbers.sort(reverse = True)
print("\nDescending Order:", numbers)

# c) length of list
print("\nLength of list is:", len(numbers))

```

```python
# question 15 - insert, delete, and display elements of a list
l = []
while True:
    print("1.insert 2.delete 3.display 4.exit")
    c = int(input("choice: "))

    if c == 1:
     x = input("element: ")
     p = int(input("position: "))
     l.insert(p,x)

    elif c == 2:
     x = input("element: ")
     if x in l:
       l.remove(x)

    elif c == 3:
       print(l)

    else:
       break
```

```python
#question 18 - commmon elements in two lists
list1 = [10,20,30,40,50]
list2 = [20,40,60,80]
common = []
for element in list1:
    if element in list2:
        common.append(element)
print("list 1: ", list1)
print("list 2: ", list2)
print("common elements: ", common)
```
