```python
# question 9
numbers = []

for i in range(20):
    numbers.append(int(input("Enter value: ")))

element = int(input("Enter element to find: "))
print("Count:", numbers.count(element))

print("Indexes:", end=" ")
for i in range(20):
    if numbers[i] == element:
        print(i, end=" ")

even = odd = positive = negative = 0

for n in numbers:
    if n % 2 == 0:
        even += 1
    else:
        odd += 1
    if n > 0:
        positive += 1
    elif n < 0:
        negative += 1

print("\nEven:", even)
print("Odd:", odd)
print("Positive:", positive)
print("Negative:", negative)
```
