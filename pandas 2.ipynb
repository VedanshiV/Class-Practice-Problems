```python
# locate row - pandas use the loc attribute to return one or more specified row(s)
# example 2
import pandas as pd
data = {
    "calories" : [420,380,390],
    "duration" : [50,40,45]
}
# load data into a dataframe object:
df = pd.DataFrame(data)
print(df.loc[0]) # loc - location
```

```python
# data cleaning and preprocessing using PANDAS
## this means fixing and organising messy data, pandas offers a wide range of tools and functions to help us clean and preprocess data efficiently
## it often involves: 1.dropping irrelevent columns 2.renaming columns to meaningful names 3.making data values consistent 4.replacing or filling in missing values
```

```python
# kaggle - (website name) data sets that are proper, consistent, and maintained 
# example 5

import pandas as pd

# define a dictionary with sample data which includes some missing values

data = {
    'A':[1,2,3,None,5],
    'B':[None,2,3,4,5],
    'C':[1,2,None,None,5]
}

df = pd.DataFrame(data)
print("Original Data:\n",df)
print()

# use dropna() [pronounced as drop NA] to remove rows with any missing values
df_cleaned = df.dropna()
print("Final Data:\n", df_cleaned)
```

```python
# fill missing values - to fill the missing values in pandas, we use the fillna() [ fill NA] function
# example 6
import pandas as pd

# define a dictionary with sample data which includes some missing values

data = {
    'A':[1,2,3,None,5],
    'B':[None,2,3,4,5],
    'C':[1,2,None,None,5]
}

df = pd.DataFrame(data)
print("Original Data:\n",df)
print()

# filling NaN values with 0
df.fillna(0, inplace = True)
print("Final Data:\n", df)
```

```python
# use aggregate values to fill missing values
## instead of filling with 0, we can use aggregate functions
## here the df.mean() calculates the mean for each column, and then the fillna() method to then replace NaN values
```

```python
# example 7
import pandas as pd

# define a dictionary with sample data which includes some missing values

data = {
    'A':[1,2,3,None,5],
    'B':[None,2,3,4,5],
    'C':[1,2,None,None,5]
}

df = pd.DataFrame(data)
print("Original Data:\n",df)
print()

#filling NaN values with the mean of each column
df.fillna(df.mean(), inplace = True)
print("Final Data:\n", df)
```

```python
# handle duplicate values 
# example 8
import pandas as pd

# define a dictionary with sample data which includes some missing values

data = {
    'A':[1,2,2,3,4,4,5],
    'B':[6,7,7,8,9,9,10]
}

df = pd.DataFrame(data)
print("Original DataFrame:\n", df.to_string(index = False))
print()

#detect duplicates
print("Duplicate Rows:\n", df[df.duplicated()].to_string(index=False)) #df[df.duplicate()]produces a boolean series tp identify duplicate rows
print()

#remove duplicates based on column 'A'

df.drop_duplicates(subset=['A'], keep = 'first', inplace = True)
# the above line removes duplicates based on column A, retaining  only the first occurance of each duplicate directly in the original DataFrame

print("DataFrame after removing duplicates based on column A:\n", df.to_string(index = False))
```

```python
# renaming column names to meaningful names
## to rename column names we use - rename() function

# example 9
import pandas as pd

#sample data
data = {
    'A' : [20,30,35],
    'B' : ['John', 'Doe', 'Smith'],
    'C' : [50000,60000,70000]
}

df = pd.DataFrame(data)

#rename columns
df.rename(columns={'A': 'Age', 'B': 'Name', 'C': 'Salary'}, inplace = True)
print(df.to_string(index=False))
```
