```python
# question 3
import math
def roots(a,b,c):
    try: 
        d = b**2 - 4*a*c
        r1 = (-b + math.sqrt(d))/ (2*a)
        r2 = (-b - math.sqrt(d))/ (2*a)
        print("roots are:", r1, "and", r2)
    except valueError:
        print("value error: no real roots")
    except nameError:
        print("name error: variable not defined")
    except typeError:
        print("type error: invalid data")
a = float(input("enter a: "))
b = float(input("enter b: "))
c = float(input("enter c: "))
roots(a,b,c)
```

```python
# recursion - when the excecutive statement is called again and again
# question 4
def print_numbers(n):
    if n == 0:
        return
    print_numbers(n-1) #recursive call
    print(n)
n = int(input("enter n: "))
print_numbers(n)
```

```python
# question 5
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n*factorial(n-1)
n = int(input("enter a number: "))
print("factorial =", factorial(n))
```

```python
# question 6
def fib(n):
    if n <= 1:
        return n
    return fib(n-1)+fib(n-2)
n = int(input("enter number of terms: "))
for i in range(n):
    print(fib(i), end = " ")
```
