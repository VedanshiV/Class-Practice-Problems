```python
# types of function arguements:
# 1.  default arguement - assumes a defalt value
def myFun(x, y = 50):
    print("x:", x)
    print("y:", y)
myFun(10)
```

```python
def my_fun(country = "norway"):
    print(" i am from " + country)
my_fun("sweden")
my_fun("india")
my_fun()
my_fun("brazil")
```

```python
# key word arguements - allow the caller to specify the arguement name with values so that the caller doesnt need to remember the order of parameters
def student(fname, lname):
    print(fname, lname)
student(fname = "good", lname = "practice")
student(lname = "practice", fname = "good")
```

```python
# positional arguements - 
def nameAge(name, age):
    print("hi, i am ",name)
    print("my name is ",age)
print("case-1: ")
nameAge("suraj",27)
print("\ncase-2: ")
nameAge(27,"suraj")
```

```python
# arbitary keyword arguements - accepts variable number of arguements and they might be positional or keyword arg
# *args = posn arg and **kwarg = arb keyword arg
def myFun(*argv):
    for arg in argv:
        print(arg)
myFun("hello", "welcome", "to", "mit")
```

```python
# recursion function - it repeats itself
def fact(n):
    if n == 0:
        return 1
    else:
        return n*fact(n-1)
print(fact(4))
```
