```python
# introduction to control structure and functions
## conditional statements and loop control statements and functions - defn and paraameters
## conditional statement - decision making 
## a) if statement : executes block of code if its true
## b) if else statement : executes one block of code if its true and another block if its false
## c) if elif else statement : allows checking multiple conditions 

# nested if statement : statements put under one another
score = 85
attendance = 92
if score >= 70:
    print("Student passed the exam.")
    if attendance >=90:
     print("student also has excelled in attendance.")
    else:
     print("student's attendance could be improved.")
else:
     print("student did not pass the exam.")
```

```python
# looping statement ( iteration ) : these statements allow a block of code to be executed repeatedly
# a) for loop : iterates over a sequence (eg list, tuple, dictionary)
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

```python
# b) while loop : a block of code as long as a condition remains true
count = 0
while count < 5:
    print(count)
    count += 1
```

```python
# c) nested loops : one or more loops inside another loop
adj = ["red","big","tasty"]
fruits = ["apple", "banana", "cherry"]
for x in adj:
    for y in fruits:
        print(x,y)
```

```python
# d) break statement : terminates current loop 
for i in range(10):
    if i == 5:
        break
    print(i)
```

```python
# e) continue statement: skips iteration of loops and proceeds on the next loop
for i in range(5):
    if i == 2:
        continue
    print(i)
```
